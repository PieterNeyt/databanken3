```python
# === Setup en verbinding ===
debugging_mode = True

from pyspark.sql.functions import col, expr
from pyspark.sql.types import StructType, StructField, StringType
import ConnectionConfig as cc
```

```python
# === Setup en verbinding ===
cc.setupEnvironment()
cc.setupEnvironment()
print(cc.config.sections())

spark = cc.startLocalCluster("DIM_SEASON")
spark.getActiveSession()
```

```python
# TRANSFORM
seasons = [("Zomer",), ("Herfst",), ("Winter",), ("Lente",)]

schema = StructType([StructField("SeasonName", StringType(), True)])

dimSeason = (
    spark.createDataFrame(seasons, schema)
         .withColumn("SeasonSurKey", expr("uuid()"))
)

if debugging_mode:
    print("Season dimensie preview:")
    dimSeason.show(truncate=False)
```

```python
# LOAD
dimSeason.coalesce(1).write.format("delta").mode("overwrite").save("delta/SEASON_DIM")

if debugging_mode:
    print("Season dimensie succesvol opgeslagen naar delta/SEASON_DIM")
```

```python
spark.stop()
```
