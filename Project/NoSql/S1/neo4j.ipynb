```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
import ConnectionConfig as cc
```

```python
#config
cc.setupEnvironment()
print(cc.config.sections())
```

```python
cc.config.read('config.ini')
cc.set_connectionProfile("catchem")

spark = cc.startLocalCluster("DIM_DATE",4)
spark.getActiveSession()

base_path = "./dataset_csv"
output_path = "./neo4j_export_csv"
```

```python
import pandas as pd

df = pd.read_csv(f"{base_path}/city.csv", header=None)
df.columns = ["city_id", "city_name", "latitude", "longitude", "postal_code", "country_code"]

# Opslaan als nieuwe CSV mét header
header_csv_path = f"{base_path}/city_with_header.csv"
df.to_csv(header_csv_path, index=False)

# Inlezen in Spark met header=True
cities_df = spark.read.option("header", True).csv(header_csv_path)

cities_df.show()
```

```python
df = pd.read_csv(f"{base_path}/hunters.csv", header=None)
df.columns = ["hunter_id", "first_name", "last_name", "email", "number", "street","city_id"]

# Opslaan als nieuwe CSV mét header
header_csv_path = f"{base_path}/hunters_with_header.csv"
df.to_csv(header_csv_path, index=False)

# Inlezen in Spark met header=True
hunters_df = spark.read.option("header", True).csv(header_csv_path)

hunters_df.show()
```

```python
df = pd.read_csv(f"{base_path}/hunts.csv", header=None)
df.columns = ["hunt_id", "description", "log_time", "log_type", "session_start", "hunter_id","treasure_id"]

# Opslaan als nieuwe CSV mét header
header_csv_path = f"{base_path}/hunts_with_header.csv"
df.to_csv(header_csv_path, index=False)

# Inlezen in Spark met header=True
hunts_df = spark.read.option("header", True).csv(header_csv_path)

hunts_df.show()
```

```python
df = pd.read_csv(f"{base_path}/treasures.csv", header=None)
df.columns = ["treasure_id", "difficulty", "terrain", "city_id", "owner_id"]

# Opslaan als nieuwe CSV mét header
header_csv_path = f"{base_path}/treasures_with_header.csv"
df.to_csv(header_csv_path, index=False)

# Inlezen in Spark met header=True
treasures_df = spark.read.option("header", True).csv(header_csv_path)

treasures_df.show()
```
